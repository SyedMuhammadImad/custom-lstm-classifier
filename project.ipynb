# recurrent classifier custom LSTM classification

A manually implemented gated LSTM cell and stacked recurrent network classify 35 ordered tabular features treated as a length-35 sequence. Cell outputs and gradients match a PyTorch LSTMCell after gate-order conversion; stacked unrolling is also checked. The feature sequence is not a time series and these results do not establish forecasting ability.


```python
import json,sys
from pathlib import Path
ROOT=next(p for p in [Path.cwd(),*Path.cwd().parents] if (p/'project.json').exists())
sys.path.insert(0,str(ROOT))
import numpy as np
import models
```

## Implementation check

This mathematical smoke check is separate from measured classifier accuracy.


```python
import torch
from experiment_core import as_tensor,torch_factories,SPECS
kind='recurrent'
factory=next(iter(torch_factories(kind).values()))[0]
model=factory().eval()
with torch.no_grad():
    probs=model(as_tensor(np.zeros((3,SPECS[kind][0])),kind)).softmax(dim=1)
assert torch.isfinite(probs).all() and torch.allclose(probs.sum(dim=1),torch.ones(3))
```
